```python
# =========================================================
# SISTEMA IA - CONTROL DE CONDUCTORES Y AUTOBUSES
# CBTIS 128
# Carrera: Inteligencia Artificial
# Materia: Elabora Proyectos de Programación Lógica
# Alumno: Nahum Flores
# NL = 21
# =========================================================

print("==============================================")
print("   IA DE CONTROL DE AUTOBUSES ESCOLARES")
print("==============================================")

# Lista vacía
conductor = []

# Diccionario vacío
losconductores = {}

while True:

    print("\n========== MENÚ PRINCIPAL ==========")
    print("1. Registrar conductor")
    print("2. Mostrar conductores")
    print("3. Buscar conductor")
    print("4. Revisar estado de autobuses")
    print("5. Salir")

    opcion = input("Selecciona una opción: ")

    # =====================================================
    # OPCIÓN 1 - REGISTRAR
    # =====================================================
    if opcion == "1":

        print("\n--- REGISTRO DE CONDUCTOR ---")

        nombre = input("Nombre del conductor: ")
        edad = input("Edad: ")
        autobus = input("Número de autobús: ")
        ruta = input("Ruta asignada: ")
        estado = input("Estado del autobús (Bueno/Regular/Malo): ")

        # Diccionario
        losconductores = {
            "nombre": nombre,
            "edad": edad,
            "autobus": autobus,
            "ruta": ruta,
            "estado": estado
        }

        # Guardar en lista
        conductor.append(losconductores)

        print("\n IA: Conductor registrado correctamente")

        # Inteligencia artificial simple
        if estado.lower() == "malo":
            print(" IA ALERTA: Autobús en mal estado")
            print(" Se recomienda mantenimiento urgente")

        elif estado.lower() == "regular":
            print(" IA AVISO: Revisar el autobús pronto")

        else:
            print(" IA: Autobús listo para trabajar")

    # =====================================================
    # OPCIÓN 2 - MOSTRAR
    # =====================================================
    elif opcion == "2":

        print("\n====== LISTA DE CONDUCTORES ======")

        if len(conductor) == 0:

            print("No hay conductores registrados")

        else:

            numero = 1

            for datos in conductor:

                print("\nConductor", numero)
                print("Nombre:", datos["nombre"])
                print("Edad:", datos["edad"])
                print("Autobús:", datos["autobus"])
                print("Ruta:", datos["ruta"])
                print("Estado:", datos["estado"])

                numero = numero + 1

    # =====================================================
    # OPCIÓN 3 - BUSCAR
    # =====================================================
    elif opcion == "3":

        buscar = input("\nEscribe el nombre del conductor: ")

        encontrado = False

        for datos in conductor:

            if datos["nombre"].lower() == buscar.lower():

                print("\n===== CONDUCTOR ENCONTRADO =====")
                print("Nombre:", datos["nombre"])
                print("Edad:", datos["edad"])
                print("Autobús:", datos["autobus"])
                print("Ruta:", datos["ruta"])
                print("Estado:", datos["estado"])

                encontrado = True

        if encontrado == False:

            print(" IA: Conductor no encontrado")

    # =====================================================
    # OPCIÓN 4 - ESTADO DE AUTOBUSES
    # =====================================================
    elif opcion == "4":

        print("\n====== REVISIÓN IA DE AUTOBUSES ======")

        if len(conductor) == 0:

            print("No existen autobuses registrados")

        else:

            for datos in conductor:

                if datos["estado"].lower() == "bueno":

                    print("\n Autobús en excelente estado")
                    print("Conductor:", datos["nombre"])
                    print("Autobús:", datos["autobus"])

                elif datos["estado"].lower() == "regular":

                    print("\n Autobús necesita revisión")
                    print("Conductor:", datos["nombre"])
                    print("Autobús:", datos["autobus"])

                else:

                    print("\n ALERTA DE IA")
                    print("Autobús fuera de servicio")
                    print("Conductor:", datos["nombre"])
                    print("Autobús:", datos["autobus"])

    # =====================================================
    # OPCIÓN 5 - SALIR
    # =====================================================
    elif opcion == "5":

        print("\n=================================")
        print(" Cerrando sistema de IA")
        print(" Gracias por usar el programa")
        print("=================================")

        break

    # =====================================================
    # OPCIÓN INVÁLIDA
    # =====================================================
    else:

        print("\n Opción inválida")
        print(" Intenta nuevamente")
```
